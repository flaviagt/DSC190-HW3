```python
#Imports
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load Dataset
df = pd.read_csv("Data/epidemiologicaldata.csv")
df.head()
```

### Part 1: Explore epidemiological data
#### Question 1.1

```python
# Convert dates so the x-axis is a real time axis
df["date"] = pd.to_datetime(df["date"])

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(df["date"], df["new_cases"], marker="o")

ax.set_title("Weekly New Plague Cases")
ax.set_xlabel("Week")
ax.set_ylabel("Number of New Cases")
ax.grid(alpha=0.3)
fig.autofmt_xdate()

plt.show()
```

#### Question 1.2

```python
# First plague case: first week with at least one new case
first_case = df.loc[df["new_cases"] > 0, "date"].min()
print("First plague case (week of):", first_case.date())
```

```python
# Last plague case: last week with at least one new case
last_case = df.loc[df["new_cases"] > 0, "date"].max()
print("Last plague case (week of):", last_case.date())
```

```python
# Peak of the outbreak: week with the largest number of new cases
peak_row = df.loc[df["new_cases"].idxmax()]
print("Outbreak peak (week of):", peak_row["date"].date(), "with", peak_row["new_cases"], "new cases")
```
